El objetivo de este cuaderno es desarrollar un programa para aproximar integrales definidas mediante el metodo de **integración de Romberg** siguiendo el algoritmo 4.1.

```python
import numpy as np
import matplotlib.pyplot as plt
```

En el siguente codigo estamso realizando el algoritomo 4.2 proporcionado del libro Burden:

```python
def romberg(f, a, b, n):
    """
    Aproximar la integral de una función f(x) en el intervalo [a,b] mediante el método de integración de Romberg.

    Parámetros:
    f: función que se desea integrar 
    a: limite inferior 
    b: limite superiro 
    n: numero de filas 

    Retotno: 
    R: tabala de Romberg
    aproximacion: valor aproximado de la integral 
    """

    if not isinstance(n, ( int, np.integer)):
        raise ValueError("n debe de ser un numero entero .")

    if n <= 0:
        raise ValueError("n debe ser mayor que cero.")

    if b<=a:
        raise ValueError("b debe ser mayor que a.")

    R = np.zeros((n, n))

    h = b - a
    R[0,0] = (h / 2) * (f(a) + f(b))
    print("R[1,1] =", R[0,0])

    for i in range(1,n):
        suma=0.0
        for k in range (1, 2**i, 2):
            x = a + k * h / (2**i)
            suma = suma + f(x)

        R[i,0] = 0.5 * R[i-1,0] + (h/ (2**i)) * suma

        for j in range (1, i + 1):
            R[i,j] = (R[i,j]+(R[i,j-1] - R[i-1, j-1] / (4**j - 1)))

    aproximacion = R[n-1, n-1]

    return R, aproximacion
```

El siguente codigo esta relacionado con la elaboracion de la tabla.

```python
def mostrar_tabla_romberg(R):
    n = len(R)
    print("\nTABLA DE RONBERG")
    print("-" *65)

    for i in range(n):

        for j in range (i + 1):
            print(f"R[{i+1},{j+1}] ={R[i,j]:.10f}", end="   ")

        print()
```

```python
def mostrar_resultados_romberg(a, b, n, resultado):

    print("\n" + "-" *70)
    print("RESULTADOS FINAL")
    print("-" * 70)

    print("Metodo de integración de Romberg")
    print("Intervalo: [",a, ",",b, "]")
    print("Números de filas n:", n)
    print("Aproximación de la integral:", resultado)
```

```python
def f(x):
    return np.sin(x)

a = 0
b = np.pi 
n = 6
```

```python
tabla, resultado = romberg(f, a, b, n)
```
