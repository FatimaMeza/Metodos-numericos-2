```python
import numpy as np
import sympy as sp
import marplotlib.pyplot as plt
```

El siguente codigo es el algoritmo 4.1 que nos proporciona el libro Burden.

```python
def simpson_compuesto(f , a, b, n):
    """
    Aproximar la integral de una funciónf(x) en el intervalo [a,b] mediante la regla compuesta de simpson

    parámetros:
    f : función que se decea integrar 
    a : límite inferiror
    b : límite superiror 
    n : número de subintervalos, positivo y par 

    Retorna: 
    XI : aproximación de la integral 
    """

    #Verificamos que n sea entero positivo y par 
    if not isinstance(n, (int, np.integer)):
        raise ValueError("n debe ser un número entero.")

    if n<=0 or n%2 !=0:
        raise ValueError("n debe de ser un entero pasitivo y par.")

    #calculamos H
    h = (b - a) / n

    XI0 = f(a) + f(b)
    XI1 = 0.0
    XI2 = 0.0

    for i in range(1, n):
        X = a + i * h

    if i%2 == 0:
        XI2 = XI2 + f(X)
    else:
        XI1 = XI1 + f(X)

    XI = h * (XI0 + 2 * XI2 + 4 * XI1) / 3

    return XI
```

En el siguente codigo realizaremos el calculo de $n$ y $h$ ya que en algunos ejercicios no los proporcionan.
Tambien relaizamos el calculo de la cuarta derivada para que el usuario no tenga que realizar ese calculo.

```python
def calcular_n_h(funcion_simbolica,variable, a, b, tolerancia):
    if tolerancia <= 0:
        raise ValueError("La tolerancia debe de ser mayor a cero.")

    if b<= a:
        raise ValueError("b debe ser mayor que a")

    cuarta_derivada = sp.diff(funcion_simbolica, variable, 4)
    f4= sp.lambdify(variable, cuarta_derivada, "numpy")

    x = np.linspace(a, b, 100000)

    M = np.max(np.abs(f4(x)))

    n_min = ((b-a)**5 * M / (180*tolerancia))**(1/4)

    n = int (np.ceil(n_min))

    if n%2 != 0:
       n = n + 1

    h = (b-a)/n

    return  n,h,M, cuarta_derivada
```
